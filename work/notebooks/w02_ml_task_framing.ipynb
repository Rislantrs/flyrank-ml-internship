## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
import pandas as pd

# URL mentah ke dataset di GitHub
github_csv_url = 'https://raw.githubusercontent.com/Rislantrs/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv'

# Memuat data untuk melihat seberapa banyak konten yang akan di-ranking
df = pd.read_csv(github_csv_url)

# Menampilkan jumlah total halaman web yang akan dievaluasi dan diberikan skor prioritas
total_content = df['content_id'].nunique()
print(f"Total konten unik yang perlu diberikan skor prioritas (di-ranking): {total_content} halaman")

# Menampilkan beberapa baris pertama data untuk pemeriksaan cepat
print("\nBeberapa baris pertama data:")
display(df.head())
```

```python
print("Daftar semua fitur (kolom) dalam dataset:")
display(df.columns.tolist())
```

The primary focus of this problem solving is to optimize content team decisions and workflows. Instead of simply classifying whether content will perform poorly (yes/no), the ranking approach will evaluate historical data over the past 90 days to generate a priority score for each webpage.

The output, in the form of a sorted content list (ranked queue), will provide actionable insights for editors: which content has the greatest impact and is most urgent to update or fix first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Memeriksa nilai apa saja yang ada di kolom trend_direction
print("Distribusi Arah Tren Konten:")
print(df['trend_direction'].value_counts(dropna=False))

# Memperbaiki nama target sesuai dengan data yang sebenarnya
target_kondisi = (df['trend_direction'] == 'down')
print(f"Jumlah konten yang menurun: {target_kondisi.sum()} halaman")
```

The target to be predicted is the probability of performance decline, which is derived from the `trend_direction` column. We convert it into a binary target: 1 if 'down' (declining) and 0 for other categories (stable, up, flat, new).

The model will output a probability score (e.g., 90% chance of being 'down'). This probability score will be our proxy for ranking the content queue, from the most urgent to be fixed.

Note: Since `trend_direction` and `trend_pct` are direct observations (ground truth), these two columns MUST be removed from the features (input) to prevent data leakage.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Menghitung Base Rate (Tebakan Acak) sebagai Baseline untuk Precision@K
total_halaman = len(df)
jumlah_down = target_kondisi.sum()

base_rate = (jumlah_down / total_halaman) * 100

print(f"Base Rate (Peluang acak halaman berstatus 'down'): {base_rate:.2f}%")
print("Artinya, jika tanpa ML kita mengambil 100 halaman secara acak, presisinya hanya ~54%.")
print("Target Metrik: Precision@100 dari model ML kita harus jauh lebih tinggi dari 54% (idealnya di atas 80-90%).")
```

The main success metric for this ranking system is Precision@K (e.g., Precision@100 or Precision@500).

Business reason: The editorial team has limited capacity. If we provide a queue of the top 100 contents (K=100) predicted to be the most urgent to fix, we want to ensure that most (close to 100%) of that list are indeed contents with declining performance.

As a secondary metric, we will use ROC-AUC to ensure the model's overall ability to separate (rank probabilities) content that is declining versus not declining. The model is considered 'good' if its Precision@K and ROC-AUC significantly outperform the base rate (random guess), which we calculated to be **54.21%** for pages with 'down' status.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Menunjukkan kompleksitas data:
# Mengabaikan aturan sederhana dengan melihat interaksi beberapa fitur terhadap status 'down'

features_to_check = ['clicks_90d', 'avg_position', 'scroll_rate', 'word_count']

print("Rata-rata metrik berdasarkan status down:")
summary = df.groupby(target_kondisi)[features_to_check].mean()
display(summary.round(2))

print("\nPenjelasan Kode:")
print("Tabel di atas menunjukkan bahwa konten yang 'down' memiliki nilai rata-rata yang saling bertumpuk dengan yang 'tidak down'.")
print("Satu aturan sederhana (misal: jika clicks_90d < X) pasti akan salah mengklasifikasikan banyak halaman. ML dibutuhkan untuk memproses gabungan seluruh sinyal ini.")
```

Using rigid rules (like if-else statements) is not sufficient to prioritize content updates because the patterns of performance decline are highly complex and involve the interaction of dozens of signals simultaneously (clicks, Google position, scroll rate, content type, and even AI search performance).

Rigid rules would impose thresholds that are too generalized for all clients, whereas the 'decline' threshold for a small blog differs from that of a corporate website. Machine Learning excels here because it can monitor non-linear interactions of 40+ features simultaneously, providing subtle (continuous) probability estimates to rank priorities much more accurately than manual if-else rules.
